```python
import pandas as pd
import pandas as pd
```

```python
data=pd.read_csv('/content/wine_quality_merged.csv')
```

```python
data.head()
```

```python
data[["type"]].value_counts()
```

```python
data.isnull().sum()
```

```python
x=data[["volatile acidity","citric acid","residual sugar","chlorides","free sulfur dioxide","total sulfur dioxide","density"]]
```

```python
x
```

```python
y=data[["type"]]
y.head()
```

```python

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42,stratify=y)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
    n_estimators=200,   # number of trees
    random_state=42,
    n_jobs=-1 )
```

```python
rf.fit(x_train, y_train)
```

```python
y_pred = rf.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report
```

```python
confusion_matrix= confusion_matrix(y_test, y_pred)
```

```python
confusion_matrix
```

```python

precision= precision_score(y_test, y_pred, average="weighted")
recall = recall_score(y_test, y_pred, average="weighted")
f1= f1_score(y_test, y_pred, average="weighted")
```

```python
precision
```

```python
recall
```

```python
f1
```

```python
import pickle


file_name=("classfication.sav")


pickle.dump(rf,open(file_name,"wb"))


result=pickle.load(open(file_name,"rb"))


```

```python
# Get user input
user_input = {}
user_input['volatile acidity'] = float(input('Enter volatile acidity: '))
user_input['citric acid'] = float(input('Enter citric acid: '))
user_input['residual sugar'] = float(input('Enter residual sugar: '))
user_input['chlorides'] = float(input('Enter chlorides: '))
user_input['free sulfur dioxide'] = float(input('Enter free sulfur dioxide: '))
user_input['total sulfur dioxide'] = float(input('Enter total sulfur dioxide: '))
user_input['density'] = float(input('Enter density: '))

# Convert to DataFrame
user_df = pd.DataFrame([user_input])

# Make prediction
prediction = result.predict(user_df)

# Display the prediction
print(f"The predicted wine type is: {prediction[0]}")
```
