```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, r2_score

scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

from sklearn.naive_bayes import MultinomialNB
mnb = MultinomialNB()
mnb.fit(x_train_scaled, y_train)
y_pred = mnb.predict(x_test_scaled)

acc = accuracy_score(y_test, y_pred)

# R² Score (not usually used for classification, but you asked)
r2 = r2_score(y_test, y_pred)

print("MultinomialNB Accuracy:", acc)
print("MultinomialNB R² Score:", r2)
```

```python
from sklearn.naive_bayes import BernoulliNB
scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

bnb = BernoulliNB()
bnb.fit(x_train_scaled, y_train)
y_pred = bnb.predict(x_test_scaled)

print("BernoulliNB Accuracy:", accuracy_score(y_test, y_pred))
print("BernoulliNB R²:", r2_score(y_test, y_pred))
```

```python
from sklearn.naive_bayes import ComplementNB

scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

cnb = ComplementNB()
cnb.fit(x_train_scaled, y_train)
y_pred = cnb.predict(x_test_scaled)

print("ComplementNB Accuracy:", accuracy_score(y_test, y_pred))
print("ComplementNB R²:", r2_score(y_test, y_pred))
```

```python
from sklearn.naive_bayes import CategoricalNB
import numpy as np

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

x_train_cat = np.round(x_train_scaled * 3).astype(int)
x_test_cat = np.round(x_test_scaled * 3).astype(int)

catnb = CategoricalNB()
catnb.fit(x_train_cat, y_train)
y_pred = catnb.predict(x_test_cat)

print("CategoricalNB Accuracy:", accuracy_score(y_test, y_pred))
print("CategoricalNB R²:", r2_score(y_test, y_pred))
```
